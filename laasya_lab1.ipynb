## 3. Variables & Data Types
Python is **dynamically typed** – variable types are inferred at runtime.

| Type | Example | Literal |
|------|---------|---------|
| `int` | whole numbers | `x = 42` |
| `float` | decimal numbers | `pi = 3.14159` |
| `str` | text | `name = "Ada"` |
| `bool` | truth values | `is_valid = True` |
| `list` | ordered, mutable collection | `scores = [95, 88, 72]` |
| `dict` | key‑value pairs | `user = {'id': 1, 'name': 'Sam'}` |

Run the cell below to see the types in action.

```python
x = 42
pi = 3.14159
name = "Ada"
is_valid = True
scores = [95, 88, 72]
user = {'id': 1, 'name': 'Sam'}

print(type(x), type(pi), type(name), type(is_valid), type(scores), type(user))
```

## Activity 3 – QuackMoo
Write a loop that prints numbers 1‑25. For multiples of 4 print "Quack", for multiples of 6 print "Moo", and for multiples of both print "QuackMoo".

```python
# Your code here
def quackMoo(num):
  for i in range(1, num + 1):
    if i % 12 == 0:
      print("quackmoo")
    elif i % 4 == 0:
      print("quack")
    elif i % 6 == 0:
      print("moo")
    else:
      print(i)

quackMoo(25)
```

## Activity 4 – Factorial Function
Implement `factorial(n)` that returns the factorial of n (`factorial(0)=1`, `factorial(1)=1`, `factorial(2)=2`, etc.). Test that `factorial(5)` is 120.

```python
# Your code here
def factorial(num):
  result = 1
  if num == 0:
    print(f"factorial(0) =  {result}")
  for i in range(1, num + 1):
    result = result * i
  print(f"factorial({num}) = {result}")


n = int(input("enter a number that you want to find the factorial of: "))
factorial(n)
```

   Implement `is_prime(n)` efficiently for `n ≤ 10^6`. Test it on the following primes: 2, 3, 5, 7, 9973

Choose any you like and work below.

```python
# prime check function

def is_prime(num):
  if num > 1000000:
    print("sorry the program won't work!")
  elif num <= 0:
    print("only positive numbers!")
  elif num == 1:
    print("1 is not a prime number")
  else:
    for i in range(2, num - 1):
      if num % i == 0:
        print(f"{num} is not a prime number")
        return
    print(f"{num} is a prime number!")

n = int(input("enter a number up till 1000000: "))
is_prime(n)
```
